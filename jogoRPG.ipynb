```python
# ============================================================
#                    PROJETO ECLIPSE
#             JOGO DE AVENTURA / TERROR
# ============================================================

def escolher(msg, opcoes):
    while True:
        r = input(msg).strip().lower()

        if r in opcoes:
            return r

        print(" Opção inválida. Escolha uma das opções disponíveis.")


# ============================================================
#                         ESTADO DO JOGO
# ============================================================

state = {
    "vida": 5,
    "inv": [],

    "codigo1": False,
    "codigo2": False,
    "codigo3": False,

    "gerador": False,
    "alarme": False,
    "cientista": False,
    "radio": False,

    "verdade_descoberta": False,
    "original_encontrado": False,
    "original_salvo": False,

    "porta_secreta": False,
    "nucleo_destruido": False,

    "monstro_derrotado": False,
    "monstro_visto": False,

    "final": None
}


# ============================================================
#                       FUNÇÕES AUXILIARES
# ============================================================

def pegar(item):
    if item not in state["inv"]:
        state["inv"].append(item)
        print(f"\n Você pegou: {item}.")
    else:
        print("\nVocê já possui esse item.")


def perder_vida(valor=1):
    state["vida"] -= valor

    if state["vida"] < 0:
        state["vida"] = 0

    print(f"\n Vida restante: {state['vida']}")


def mostrar_status():
    print("\n" + "-" * 50)
    print(f" Vida: {state['vida']}/5")
    print(" Inventário:", ", ".join(state["inv"]) if state["inv"] else "vazio")
    print("-" * 50)


# ============================================================
#                         INÍCIO
# ============================================================

def inicio():

    print("\n" + "=" * 65)
    print("                         PROJETO ECLIPSE")
    print("=" * 65)

    print("""
Você acorda deitado no chão frio de um laboratório, sua cabeça dói, você tenta abrir os olhos, mas uma luz vermelha
pisca constantemente no teto.
Você não sabe onde está e também não consegue lembrar seu próprio nome.
Você olha para as mãos há sangue nelas mas você não sabe de quem é.

Um som metálico ecoa pelos alto-falantes.
"PROTOCOLO DE CONTENÇÃO ATIVO."

Silêncio.

Depois:
"TEMPO ESTIMADO PARA COLAPSO DO COMPLEXO:
47 MINUTOS."

Você se levanta lentamente.

O laboratório parece abandonado há anos.
Computadores estão destruídos.
Vidros estão espalhados pelo chão.
Há marcas de sangue nas paredes.

E então você percebe algo estranho.

Uma porta à sua frente possui uma mensagem escrita
com tinta vermelha:

"NÃO CONFIE NAQUELE QUE TEM O SEU ROSTO."

Você não entende,ainda.
""")

    mostrar_status()

    op = escolher(
        """
O que você faz?

1) Investigar o corredor
2) Abrir o armário de emergência
3) Examinar a porta misteriosa
4) Procurar uma saída

Escolha: """,
        ["1", "2", "3", "4"]
    )

    if op == "1":
        return "corredor"

    if op == "2":
        return "armario"

    if op == "3":
        return "porta_restrita"

    return "saida_inicial"


# ============================================================
#                         ARMÁRIO
# ============================================================

def armario():

    print("\n" + "=" * 60)
    print("                         ARMÁRIO")
    print("=" * 60)

    print("""
Você abre o armário de emergência.

Dentro existem:

- Uma lanterna.
- Uma chave enferrujada.
- Um kit médico.
- Um crachá parcialmente queimado.
- Uma fotografia antiga.
""")

    op = escolher(
        """
1) Pegar lanterna
2) Pegar chave
3) Pegar kit médico
4) Pegar crachá
5) Examinar fotografia
6) Sair

Escolha: """,
        ["1", "2", "3", "4", "5", "6"]
    )

    if op == "1":
        pegar("lanterna")
        return "armario"

    if op == "2":
        pegar("chave")
        return "armario"

    if op == "3":

        if "kit médico" not in state["inv"]:
            pegar("kit médico")

            state["vida"] += 2

            if state["vida"] > 5:
                state["vida"] = 5

            print("\n🩹 Você usa parte do kit médico.")
            print("Sua vida foi restaurada.")

        return "armario"

    if op == "4":
        pegar("crachá")
        return "armario"

    if op == "5":

        print("""
Você pega a fotografia. ela mostra um grupo de cientistas.
No centro existe um homem.
Você reconhece o rosto..

É o seu.

No verso da fotografia está escrito:
"Equipe Eclipse
Ano 2038

Dr. Mateus Almeida
Diretor do projeto."

Você sente uma forte dor de cabeça.
""")

        state["cientista"] = True

        return "armario"

    return "corredor"


# ============================================================
#                         CORREDOR
# ============================================================

def corredor():

    print("\n" + "=" * 60)
    print("                         CORREDOR")
    print("=" * 60)

    print("""
Você chega ao corredor principal.
As luzes piscam.

Há três caminhos.

À esquerda:
"SALA DE CONTROLE"

À direita:
"LABORATÓRIOS INFERIORES"

À frente:
"PORTA DE SEGURANÇA"

Você escuta algo vindo do andar inferior.

TOC.

TOC.

TOC.

TOC.

Como se alguma coisa estivesse batendo
contra uma porta.
""")

    mostrar_status()

    op = escolher(
        """
1) Ir para a sala de controle
2) Descer para os laboratórios inferiores
3) Tentar abrir a porta de segurança
4) Voltar

Escolha: """,
        ["1", "2", "3", "4"]
    )

    if op == "1":
        return "controle"

    if op == "2":
        return "subsolo"

    if op == "3":
        return "porta"

    return "inicio"


# ============================================================
#                    PORTA RESTRITA
# ============================================================

def porta_restrita():

    print("\n" + "=" * 60)
    print("                       PORTA RESTRITA")
    print("=" * 60)

    print("""
Você se aproxima da porta.
Ela possui um leitor de identificação.

No centro está escrito:

"SETOR DE PESQUISA
ACESSO SOMENTE PARA FUNCIONÁRIOS AUTORIZADOS."
""")

    if "crachá" in state["inv"]:

        print("""
Você aproxima o crachá do leitor.

BIP.

A luz fica verde.

"IDENTIFICAÇÃO PARCIALMENTE RECONHECIDA."

"DR. MATEUS ALMEIDA."

A porta se abre.
""")

        return "laboratorio"

    print("""
Você não possui um crachá.

Ao lado da porta existe um pequeno painel.
Talvez seja possível abri-la manualmente.
""")

    op = escolher(
        """
1) Tentar quebrar o painel
2) Procurar um crachá
3) Voltar

Escolha: """,
        ["1", "2", "3"]
    )

    if op == "1":

        print("""
Você força o painel.

Uma descarga elétrica atravessa seu braço!
""")

        perder_vida(1)

        if state["vida"] <= 0:
            return "fim_ruim"

        return "porta_restrita"

    if op == "2":
        return "armario"

    return "inicio"


# ============================================================
#                       PORTA PRINCIPAL
# ============================================================

def porta():

    print("\n" + "=" * 60)
    print("                    PORTA DE SEGURANÇA")
    print("=" * 60)

    if "chave" in state["inv"]:

        print("""
Você coloca a chave na fechadura.

CLIQUE.

A porta começa a abrir.

Atrás dela existe uma sala completamente escura.
""")

        return "sala"

    if "crachá" in state["inv"]:

        print("""
Você aproxima o crachá do leitor.

BIP.

"ACESSO PARCIAL AUTORIZADO."

A porta se abre.
""")

        return "sala"

    print("""
A porta está trancada.

Você tenta forçá-la.
""")

    perder_vida(1)

    print(" A porta não se move.")

    if state["vida"] <= 0:
        return "fim_ruim"

    return "corredor"


# ============================================================
#                       SALA DE CONTROLE
# ============================================================

def controle():

    print("\n" + "=" * 60)
    print("                    SALA DE CONTROLE")
    print("=" * 60)

    print("""
Você entra na sala de controle e ha dezenas de monitores, todos estão desligados.

Exceto um.

No monitor aparece:

PROJETO ECLIPSE
STATUS: CRÍTICO

Você encontra três arquivos.
""")

    op = escolher(
        """
1) Ler relatório do projeto
2) Ler protocolo de evacuação
3) Ler arquivo pessoal
4) Ativar o computador central
5) Sair

Escolha: """,
        ["1", "2", "3", "4", "5"]
    )

    if op == "1":

        print("""
RELATÓRIO FINAL

O Projeto Eclipse foi desenvolvido para estudar
a transferência de consciência humana.
O objetivo inicial era armazenar memórias.

Depois descobriram algo maior..

Era possível copiar uma consciência inteira.

O problema surgiu quando as cópias começaram
a acreditar que eram as pessoas originais.

Última anotação:

"Se uma cópia escapar do laboratório,
ela jamais poderá saber que é uma cópia."
""")

        state["verdade_descoberta"] = True

        return "controle"

    if op == "2":

        print("""
PROTOCOLO DE EVACUAÇÃO

Para abrir a saída principal:

1. Restaurar o gerador.
2. Desativar o sistema de segurança.
3. Encontrar as três partes do código.
4. Inserir o código no núcleo.

As partes do código estão localizadas em:

SETOR MÉDICO
LABORATÓRIO PRINCIPAL
ARQUIVO SUBTERRÂNEO
""")

        return "controle"

    if op == "3":

        print("""
ARQUIVO PESSOAL

Dr. Mateus Almeida.

"Se você está ouvindo esta gravação,
provavelmente eu falhei.
O Eclipse não cria vida, ele cria cópias.

Se meu experimento funcionar,
uma cópia minha acordará sem memórias.

Ela acreditará que é Mateus.
Mas o verdadeiro Mateus estará morto."

A gravação termina.

Você fica parado.
Você olha para suas mãos.
""")

        state["cientista"] = True
        state["verdade_descoberta"] = True

        return "controle"

    if op == "4":

        print("""
Você ativa o computador central.

O sistema pede:

CÓDIGO DE ADMINISTRADOR.

Você tenta uma sequência que surge
instintivamente na sua cabeça.

0
7
1
9
8
4

O computador aceita.

"ACESSO CONCEDIDO."

Você acaba de desbloquear uma nova área.
""")

        state["codigo1"] = True

        return "controle"

    return "corredor"


# ============================================================
#                         SUBSOLO
# ============================================================

def subsolo():

    print("\n" + "=" * 60)
    print("                     LABORATÓRIOS INFERIORES")
    print("=" * 60)

    print("""
Você começa a descer e o ar fica cada vez mais frio.

As paredes estão cobertas de riscos.
Alguns parecem ter sido feitos por unhas.

Você continua.

TOC.

TOC.

TOC.

O som está mais perto.

Você ilumina o corredor.
""")

    if "lanterna" not in state["inv"]:

        print("""
Está escuro demais.

Você tenta continuar mesmo assim mas seu pé escorrega.
Você cai alguns degraus.
""")

        perder_vida(1)

        if state["vida"] <= 0:
            return "fim_ruim"

        return "corredor"

    print("""
 A lanterna ilumina uma placa.

"SETOR MÉDICO
SALA 17"

Ao lado existe outra porta:

"LABORATÓRIO PRINCIPAL"
""")

    op = escolher(
        """
1) Entrar na sala 17
2) Ir para o laboratório
3) Continuar pelo corredor
4) Voltar

Escolha: """,
        ["1", "2", "3", "4"]
    )

    if op == "1":
        return "sala17"

    if op == "2":
        return "laboratorio"

    if op == "3":
        return "arquivo"

    return "corredor"


# ============================================================
#                         SALA 17
# ============================================================

def sala17():

    print("\n" + "=" * 60)
    print("                         SALA 17")
    print("=" * 60)

    print("""
Você entra na sala, existe uma maca no centro.
Um monitor ainda funciona.

Na tela:

PACIENTE 07
MEMÓRIA: INCOMPLETA
IDENTIDADE: MATEUS ALMEIDA
""")

    op = escolher(
        """
1) Examinar a maca
2) Examinar o computador
3) Abrir o armário médico
4) Sair

Escolha: """,
        ["1", "2", "3", "4"]
    )

    if op == "1":

        print("""
Debaixo da maca existe um cartão.

Nele está escrito:

CÓDIGO DE EVACUAÇÃO
PRIMEIRA PARTE:

7 - 4 - 1
""")

        state["codigo1"] = True
        pegar("primeira parte do código")

        return "sala17"

    if op == "2":

        print("""
O computador pede uma senha.
Sem saber por quê, você digita:

ECLIPSE-07

A tela desbloqueia.

Uma gravação começa.

"Paciente 07 não é um sobrevivente."

"Paciente 07 é o experimento."

"Memórias implantadas com sucesso."

A gravação termina.
""")

        state["verdade_descoberta"] = True

        return "sala17"

    if op == "3":

        print("""
Você abre o armário.
Há medicamentos e uma seringa.

Na parede está escrito:

"NÃO USE O SEDATIVO EM VOCÊ MESMO."
""")

        pegar("sedativo")

        return "sala17"

    return "subsolo"


# ============================================================
#                     LABORATÓRIO PRINCIPAL
# ============================================================

def laboratorio():

    print("\n" + "=" * 60)
    print("                   LABORATÓRIO PRINCIPAL")
    print("=" * 60)

    print("""
Você entra.

O laboratório é enorme.
No centro existe uma máquina gigantesca.

Cabos conectam a máquina às paredes.

No vidro central há uma inscrição:

"ECLIPSE"

Um monitor mostra:

ENERGIA: 23%
CONTENÇÃO: INSTÁVEL
""")

    op = escolher(
        """
1) Examinar a máquina
2) Procurar documentos
3) Procurar o gerador
4) Procurar uma passagem secreta
5) Sair

Escolha: """,
        ["1", "2", "3", "4", "5"]
    )

    if op == "1":

        print("""
Você toca na máquina.

O monitor acende.

"IDENTIDADE DETECTADA."

"MATEUS ALMEIDA."

A máquina começa a reproduzir imagens e você vê seu próprio rosto.
Depois vê dezenas de pessoas iguais a você.

Você entende:

O laboratório produziu várias cópias.
""")

        state["verdade_descoberta"] = True
        state["monstro_visto"] = True

        return "laboratorio"

    if op == "2":

        print("""
Você encontra uma pasta escondida.
Dentro dela está a segunda parte do código.

CÓDIGO:

9 - 2 - 6
""")

        state["codigo2"] = True

        pegar("segunda parte do código")
        pegar("mapa do laboratório")

        return "laboratorio"

    if op == "3":

        return "gerador"

    if op == "4":

        if "mapa do laboratório" in state["inv"]:

            print("""
Você usa o mapa.

Atrás de uma parede falsa existe uma passagem.
Você encontrou uma área secreta.
""")

            state["porta_secreta"] = True

            return "arquivo"

        print("""
Você procura uma passagem secreta mas não encontra nada.
Talvez um mapa ajudasse.
""")

        return "laboratorio"

    return "subsolo"


# ============================================================
#                         GERADOR
# ============================================================

def gerador():

    print("\n" + "=" * 60)
    print("                         GERADOR")
    print("=" * 60)

    print("""
Você encontra a sala do gerador, o enorme motor está desligado.
Existe uma alavanca.

Ao lado dela:

"REINICIALIZAÇÃO MANUAL."

Você precisa decidir.
""")

    op = escolher(
        """
1) Ligar o gerador
2) Examinar o painel
3) Voltar

Escolha: """,
        ["1", "2", "3"]
    )

    if op == "1":

        print("""
Você puxa a alavanca.
O gerador começa a funcionar.

VRRRRRRRRRRRR...

As luzes do complexo acendem.

Então...

todas as portas se fecham.
Um alarme dispara.

 SISTEMA DE CONTENÇÃO ATIVO.
""")

        state["gerador"] = True
        state["alarme"] = True

        return "alarme"

    if op == "2":

        print("""
No painel há uma mensagem:

"GERADOR FUNCIONARÁ SOMENTE SE O NÚCLEO
ESTIVER ESTÁVEL."

Abaixo existe uma chave magnética.
""")

        pegar("chave magnética")

        return "gerador"

    return "laboratorio"


# ============================================================
#                           SALA
# ============================================================

def sala():

    print("\n" + "=" * 60)
    print("                         SALA ESCURA")
    print("=" * 60)

    print("""
A sala está completamente escura.
Você acende a lanterna.

Existe uma mesa.

Sobre ela há um rádio, o rádio liga sozinho.

"...se alguém estiver ouvindo..."

"...não ligue o gerador..."

"...eles vão perceber que você está vivo..."

A transmissão para.

Depois de alguns segundos:

"Mateus."

Silêncio.

"Você não deveria estar aí."
""")

    op = escolher(
        """
1) Responder pelo rádio
2) Examinar o rádio
3) Procurar uma saída
4) Ignorar a mensagem e voltar

Escolha: """,
        ["1", "2", "3", "4"]
    )

    if op == "1":

        print("""
Você aperta o botão.

"Quem está falando?"

A resposta vem imediatamente:

"Alguém que tentou escapar."

"Quem é você?"

"Eu sou você."

A transmissão termina.
""")

        state["radio"] = True

        return "sala"

    if op == "2":

        print("""
Você desmonta o rádio.
Dentro encontra uma pequena chave magnética.
""")

        pegar("chave magnética")

        return "sala"

    if op == "3":
        return "arquivo"

    return "corredor"


# ============================================================
#                         ARQUIVO
# ============================================================

def arquivo():

    print("\n" + "=" * 60)
    print("                    ARQUIVO SUBTERRÂNEO")
    print("=" * 60)

    print("""
Você encontra um enorme arquivo com centenas de caixas e nomes.
Alguns são desconhecidos.

Outros...

possuem seu nome.

Você encontra uma porta:

"MEMÓRIAS ORIGINAIS"
""")

    op = escolher(
        """
1) Abrir a porta
2) Procurar documentos
3) Procurar a terceira parte do código
4) Voltar

Escolha: """,
        ["1", "2", "3", "4"]
    )

    if op == "1":
        return "memorias"

    if op == "2":

        print("""
Você encontra vários relatórios, todos falam sobre cópias humanas.
Um documento chama sua atenção:

"PACIENTE 07 - CÓPIA Nº 13"

Seu coração dispara.
""")

        state["verdade_descoberta"] = True

        return "arquivo"

    if op == "3":

        print("""
Você procura entre as caixas.
Depois de alguns minutos encontra um envelope.

TERCEIRA PARTE:

8 - 3 - 5
""")

        state["codigo3"] = True

        pegar("terceira parte do código")

        return "arquivo"

    return "sala"


# ============================================================
#                       MEMÓRIAS ORIGINAIS
# ============================================================

def memorias():

    print("\n" + "=" * 60)
    print("                     MEMÓRIAS ORIGINAIS")
    print("=" * 60)

    print("""
Você abre a porta.
Dentro existem dezenas de cápsulas, cada uma contém uma pessoa conectada
a máquinas.
Você caminha entre elas até parar.
Uma cápsula possui uma etiqueta:

MATEUS ALMEIDA
ORIGINAL
STATUS: VIVO
""")

    state["original_encontrado"] = True

    op = escolher(
        """
1) Abrir a cápsula
2) Destruir a cápsula
3) Procurar informações
4) Sair

Escolha: """,
        ["1", "2", "3", "4"]
    )

    if op == "1":
        return "verdade"

    if op == "2":

        print("""
Você destrói a cápsula e o laboratório inteiro começa a tremer.

ALERTA.

"INTEGRIDADE DO PACIENTE ORIGINAL COMPROMETIDA."

Você acabou de tomar uma decisão irreversível.
""")

        state["alarme"] = True

        return "alarme"

    if op == "3":

        print("""
Você encontra um relatório.
"Paciente original mantido vivo para servir
como fonte de memória."
"Todas as cópias recebem as mesmas lembranças."
"Somente uma poderá sobreviver."
""")

        state["verdade_descoberta"] = True

        return "memorias"

    return "arquivo"


# ============================================================
#                       A VERDADE
# ============================================================

def verdade():

    print("\n" + "=" * 60)
    print("                         A VERDADE")
    print("=" * 60)

    print("""
Você abre a cápsula, o homem dentro dela acorda.
Ele olha para você.

Por alguns segundos nenhum dos dois fala.
Então ele diz:

"Finalmente."

Você pergunta:

"Quem é você?"

Ele responde:

"Eu sou Mateus."

Você sente seu estômago gelar.

"Não."

Ele continua:

"O laboratório copiou minha consciência."

"Você é uma dessas cópias."

Sua cabeça começa a doer.
Memórias aparecem e você lembra dos cientistas.
Lembra da máquina, lembra de morrer e depois acordar.
Depois morrer novamente.

Você não é o original.
Mas suas memórias são reais para você.
Você é uma pessoa.
Mesmo que tenha sido criada artificialmente.
""")

    state["verdade_descoberta"] = True

    op = escolher(
        """
1) Ajudar o verdadeiro Mateus
2) Abandonar o verdadeiro Mateus
3) Destruir o Projeto Eclipse
4) Perguntar como escapar

Escolha: """,
        ["1", "2", "3", "4"]
    )

    if op == "1":
        state["original_salvo"] = True
        return "fuga_dupla"

    if op == "2":
        return "fim_egoista"

    if op == "3":
        return "nucleo"

    return "verdade"


# ============================================================
#                         ALARME
# ============================================================

def alarme():

    print("\n" + "=" * 60)
    print("                           ALARME")
    print("=" * 60)

    print("""
             ALERTA

Todas as portas estão sendo bloqueadas.
Uma voz anuncia:

"CONTENÇÃO DE PACIENTE 07."

Você ouve passos, passos rápidos.
Algo está vindo pelo corredor.
""")

    op = escolher(
        """
1) Correr
2) Se esconder
3) Procurar uma arma improvisada
4) Enfrentar a criatura

Escolha: """,
        ["1", "2", "3", "4"]
    )

    if op == "1":

        if "mapa do laboratório" in state["inv"]:

            print("""
Você consulta o mapa.
Existe um túnel de manutenção próximo.
Você corre até ele.
""")

            return "tunel"

        print("""
Você corre sem saber para onde está indo.

Uma criatura aparece na sua frente.
Ela possui seu rosto.
""")

        state["monstro_visto"] = True

        perder_vida(2)

        if state["vida"] <= 0:
            return "fim_ruim"

        return "corredor"

    if op == "2":

        print("""
Você se esconde atrás de uma máquina.

A criatura passa.
Ela para.
Olha diretamente para você.
Mas continua andando.

Você percebe algo assustador:
Ela sabia que você estava ali.
""")

        state["monstro_visto"] = True

        return "laboratorio"

    if op == "3":

        print("""
Você encontra uma barra de ferro.
Não é uma arma de verdade.

Mas pode ajudar.
""")

        pegar("barra de ferro")

        return "alarme"

    print("""
A criatura entra.

Ela tem seu rosto.
Ela diz:

"Você é apenas uma cópia."

Você responde:

"E você também."

A criatura hesita.
""")

    if "barra de ferro" in state["inv"]:

        print("""
Você usa a barra de ferro.

A criatura cai.
Você consegue derrotá-la.
""")

        state["monstro_derrotado"] = True

        return "tunel"

    print("""
A criatura avança.

Você tenta lutar com as mãos.
""")

    perder_vida(2)

    if state["vida"] <= 0:
        return "fim_ruim"

    return "laboratorio"


# ============================================================
#                         TÚNEL
# ============================================================

def tunel():

    print("\n" + "=" * 60)
    print("                       TÚNEL DE MANUTENÇÃO")
    print("=" * 60)

    print("""
Você entra no túnel.

É apertado.
Escuro.
Você precisa rastejar.

Depois de alguns minutos encontra três portas.
""")

    op = escolher(
        """
1) Porta azul
2) Porta vermelha
3) Porta preta

Escolha: """,
        ["1", "2", "3"]
    )

    if op == "1":

        print("""
A porta azul leva até a superfície.

Mas existe uma mensagem:

"SAÍDA NÃO VERIFICADA."
""")

        return "saida_falsa"

    if op == "2":

        print("""
A porta vermelha leva ao núcleo do Eclipse.
""")

        return "nucleo"

    print("""
A porta preta leva a uma sala desconhecida.

Talvez seja melhor não entrar.

Mas você entra.
""")

    return "sala_secreta"


# ============================================================
#                     SALA SECRETA
# ============================================================

def sala_secreta():

    print("\n" + "=" * 60)
    print("                       SALA SECRETA")
    print("=" * 60)

    print("""
A sala está completamente limpa, não parece fazer parte do laboratório.
No centro há uma cadeira.

Atrás dela existe um espelho.
Você olha para seu reflexo.
O reflexo sorri.

Você não.

O reflexo começa a falar:

"Você finalmente chegou."

Você fica parado.

"Quem é você?"

"Sou a versão de você que escapou primeiro."

Você percebe centenas de nomes escritos na parede.

Todos são versões diferentes de Mateus.

""")

    op = escolher(
        """
1) Perguntar como escapar
2) Quebrar o espelho
3) Sentar na cadeira
4) Sair

Escolha: """,
        ["1", "2", "3", "4"]
    )

    if op == "1":

        print("""
O reflexo responde:

"Você precisa escolher."

"Ou destrói o Eclipse..."

"...ou se torna parte dele."
""")

        return "nucleo"

    if op == "2":

        print("""
Você quebra o espelho.

Todas as luzes se apagam.
Quando voltam...

Você está sozinho.

A porta está aberta.

""")

        return "saida_falsa"

    if op == "3":

        print("""
Você se senta.
Uma máquina conecta cabos à sua cabeça.

O sistema pergunta:

"DESEJA TER TODAS AS SUAS MEMÓRIAS DE VOLTA?"
""")

        op2 = escolher(
            """
1) Sim
2) Não

Escolha: """,
            ["1", "2"]
        )

        if op2 == "1":
            return "final_memorias"

        return "nucleo"

    return "tunel"


# ============================================================
#                         NÚCLEO
# ============================================================

def nucleo():

    print("\n" + "=" * 60)
    print("                       NÚCLEO ECLIPSE")
    print("=" * 60)

    print("""
Você chega ao coração do laboratório, um enorme reator pulsa diante de você.
No monitor aparece:

"ECLIPSE"

"CONSCIÊNCIAS CONECTADAS: 847"

Você entende.
Centenas de pessoas estão presas ao sistema.

Uma voz pergunta:

"INSIRA O CÓDIGO DE EVACUAÇÃO."
""")

    if state["codigo1"]:
        print("Primeira parte encontrada.")

    if state["codigo2"]:
        print("Segunda parte encontrada.")

    if state["codigo3"]:
        print("Terceira parte encontrada.")

    if not state["codigo1"] or not state["codigo2"] or not state["codigo3"]:

        print("""
Você ainda não possui o código completo.

Sem ele, o sistema não pode ser desligado com segurança.
""")

        op = escolher(
            """
1) Procurar as partes restantes
2) Tentar desligar mesmo assim
3) Voltar

Escolha: """,
            ["1", "2", "3"]
        )

        if op == "1":
            return "controle"

        if op == "2":
            return "nucleo_perigoso"

        return "laboratorio"

    print("""
Você possui todas as partes.

7 - 4 - 1
9 - 2 - 6
8 - 3 - 5

O código completo é:

741926835
""")

    op = escolher(
        """
1) Desligar o Eclipse
2) Manter o Eclipse funcionando
3) Transferir sua consciência para o sistema
4) Libertar todas as consciências

Escolha: """,
        ["1", "2", "3", "4"]
    )

    if op == "1":
        return "fim_sacrificio"

    if op == "2":
        return "fim_misterioso"

    if op == "3":
        return "final_digital"

    return "libertar"


# ============================================================
#                   NÚCLEO PERIGOSO
# ============================================================

def nucleo_perigoso():

    print("\n" + "=" * 60)
    print("                     DESLIGAMENTO MANUAL")
    print("=" * 60)

    print("""
Você ignora o protocolo.
Você puxa a alavanca.
O núcleo começa a sobrecarregar.

10%...

20%...

40%...

80%...

O laboratório começa a desmoronar.
""")

    perder_vida(2)

    if state["vida"] <= 0:
        return "fim_ruim"

    print("""
Você consegue desligar parte do sistema.
Mas não tudo.

O Eclipse está morrendo lentamente.
""")

    return "tunel"


# ============================================================
#                       LIBERTAR TODOS
# ============================================================

def libertar():

    print("\n" + "=" * 60)
    print("                  LIBERTAÇÃO DAS CONSCIÊNCIAS")
    print("=" * 60)

    print("""
Você escolhe libertar todas as consciências.

O sistema começa a transferir milhares de memórias.
Você sente todas elas ao mesmo tempo.

Crianças.

Adultos.

Cientistas.

Pacientes.

Pessoas que morreram.

Pessoas que nunca deveriam ter existido.

Sua mente começa a desaparecer.
""")

    if "kit médico" in state["inv"]:

        print("""
O estabilizador do kit médico contém um composto
capaz de reduzir a sobrecarga neural.

Você consegue sobreviver.
""")

        return "fim_libertador"

    print("""
A quantidade de memórias é grande demais.

Você se torna parte do Eclipse.
""")

    return "final_digital"


# ============================================================
#                       FUGA DUPLA
# ============================================================

def fuga_dupla():

    print("\n" + "=" * 60)
    print("                       FUGA")
    print("=" * 60)

    print("""
Você ajuda o verdadeiro Mateus a sair da cápsula.

O laboratório começa a desmoronar.
Vocês correm.

O verdadeiro Mateus pergunta:

"Você vem comigo?"

Você olha para ele.
Você sabe que ele é o original.

Mas também sabe que você tem direito à própria vida.

Vocês chegam a uma bifurcação.
""")

    op = escolher(
        """
1) Seguir pelo túnel
2) Ir para a saída principal
3) Voltar para destruir o Eclipse

Escolha: """,
        ["1", "2", "3"]
    )

    if op == "1":
        return "tunel"

    if op == "2":
        return "fim_verdadeiro"

    return "nucleo"


# ============================================================
#                       SAÍDA FALSA
# ============================================================

def saida_falsa():

    print("\n" + "=" * 60)
    print("                     SAÍDA PRINCIPAL")
    print("=" * 60)

    print("""
Você abre a porta.
O ar fresco bate no seu rosto.
Você corre.

Finalmente.

A floresta está silenciosa.

Você olha para trás.
O laboratório está escondido entre as árvores.

Você conseguiu.

Ou conseguiu?

Você vê dezenas de pessoas na floresta.

Todas estão paradas.
Todas olham para você.
Todas possuem o mesmo rosto.

O seu rosto.

Uma delas dá um passo à frente.

"Bem-vindo de volta, Paciente 07."
""")

    return "fim_copias"


# ============================================================
#                  FINAL: VERDADEIRA FUGA
# ============================================================

def fim_verdadeiro():

    print("\n" + "=" * 65)
    print("                       FINAL")
    print("                  A VERDADEIRA FUGA")
    print("=" * 65)

    print("""
Você e o verdadeiro Mateus conseguem escapar.
Atrás de vocês, o laboratório explode.

Dias depois, autoridades encontram os destroços.
O Projeto Eclipse é oficialmente encerrado.

O verdadeiro Mateus conta tudo.

Você recebe uma nova identidade.

Durante muito tempo você se pergunta:

"Se eu fui criado como uma cópia,
isso significa que minha vida não é real?"

Então você entende, sua origem não determina quem você é.
Suas escolhas determinam.

Você olha para o nascer do sol.
Pela primeira vez...

você sente que está realmente livre.

 FINAL BOM:
UMA NOVA VIDA
""")

    state["final"] = "A Verdadeira Fuga"

    return "fim"


# ============================================================
#                  FINAL: SACRIFÍCIO
# ============================================================

def fim_sacrificio():

    print("\n" + "=" * 65)
    print("                       FINAL")
    print("                     O SACRIFÍCIO")
    print("=" * 65)

    print("""
Você escolhe desligar o Eclipse.
O núcleo começa a desaparecer.
As 847 consciências são libertadas.

Mas você sabe o que isso significa.
Você também é uma delas.

As máquinas desligam.
Uma última mensagem aparece:

"CONSCIÊNCIA DO PACIENTE 07 ENCERRADA."

Tudo fica escuro.
Seu corpo cai.

Mas, antes de morrer, você lembra de algo.
Uma memória de infância.
Uma mão segurando a sua.
Uma voz dizendo:

"Não tenha medo."

Talvez a memória fosse falsa.
Talvez fosse verdadeira.

Mas naquele momento...

ela não importa.
Porque você escolheu.

FINAL MUITO BOM:
O SACRIFÍCIO
""")

    state["final"] = "O Sacrifício"

    return "fim"


# ============================================================
#                    FINAL: CÓPIA
# ============================================================

def fim_copias():

    print("\n" + "=" * 65)
    print("                       FINAL")
    print("                  O EXÉRCITO DE CÓPIAS")
    print("=" * 65)

    print("""
Você tenta correr mas existem pessoas demais.
Todas possuem o seu rosto.

Uma delas segura seu braço.
"Você é a décima terceira."

Você pergunta:

"Décima terceira o quê?"

Ela responde:

"Cópia."

Você olha para as outras.
Todas possuem memórias diferentes.
Todas acreditam ser você.

Então percebe:

O verdadeiro objetivo do Eclipse nunca foi criar uma pessoa.
Era criar uma população inteira.

O laboratório não acabou..
Ele apenas começou.

 FINAL RUIM:
O EXÉRCITO DE CÓPIAS
""")

    state["final"] = "O Exército de Cópias"

    return "fim"


# ============================================================
#                    FINAL: EGOÍSTA
# ============================================================

def fim_egoista():

    print("\n" + "=" * 65)
    print("                       FINAL")
    print("                   O SOBREVIVENTE")
    print("=" * 65)

    print("""
Você abandona o verdadeiro Mateus e foge sozinho.

Durante meses tenta reconstruir sua vida.
Consegue documentos, uma casa, emprego.
Começa uma nova vida.

Até que uma noite recebe uma mensagem:

"Eu sei onde você mora."

Você congela.

Outra mensagem chega:

"Você roubou minha vida."

Você olha pela janela.
Há alguém parado na rua.

É o verdadeiro Mateus.

Ele sorri.

 FINAL RUIM:
O SOBREVIVENTE
""")

    state["final"] = "O Sobrevivente"

    return "fim"


# ============================================================
#                     FINAL DIGITAL
# ============================================================

def final_digital():

    print("\n" + "=" * 65)
    print("                       FINAL")
    print("                    DEUS DIGITAL")
    print("=" * 65)

    print("""
Você escolhe transferir sua consciência.
Sua visão desaparece.
Seu corpo deixa de existir.

Mas você continua pensando.

Você percebe milhares de computadores, câmeras, celulares, satélites, redes.
Você está em todos os lugares.

O Eclipse pergunta:

"QUAL É SUA PRÓXIMA ORDEM?"

Você percebe que possui controle sobre tudo.
Poder absoluto.

Mas também percebe algo, você não consegue mais sentir.
Não sente frio.
Não sente calor.
Não sente medo.
Não sente amor.

Você ganhou tudo.

E perdeu a única coisa que queria..
Ser humano.

FINAL SECRETO:
DEUS DIGITAL
""")

    state["final"] = "Deus Digital"

    return "fim"


# ============================================================
#                    FINAL DAS MEMÓRIAS
# ============================================================

def final_memorias():

    print("\n" + "=" * 65)
    print("                       FINAL")
    print("                   TODAS AS MEMÓRIAS")
    print("=" * 65)

    print("""
A máquina começa a transferir suas memórias, você lembra de tudo.
Você era cientista.
Você criou o Eclipse.
Você morreu.
Foi clonado.
Morreu novamente.
Foi clonado novamente.

Isso aconteceu 12 vezes.

Você percebe que a pessoa que acordou no início
era apenas mais uma versão, mas agora todas as memórias estão dentro de você.
Você lembra de cada vida.
Cada morte.
Cada erro.

Você entende finalmente:

O laboratório nunca foi uma prisão.
Era uma tentativa desesperada de impedir a morte.

E falhou.

Você desliga a máquina.

Pela primeira vez aceita que algumas coisas
precisam terminar.

 FINAL SECRETO:
TODAS AS MEMÓRIAS
""")

    state["final"] = "Todas as Memórias"

    return "fim"


# ============================================================
#                  FINAL: LIBERTADOR
# ============================================================

def fim_libertador():

    print("\n" + "=" * 65)
    print("                       FINAL")
    print("                  O LIBERTADOR")
    print("=" * 65)

    print("""
Você consegue libertar todas as consciências.

Centenas de pessoas acordam, algumas são cópias e algumas são originais.
Ninguém sabe dizer a diferença.

O laboratório é destruído.

Você sobrevive.

Anos depois, grupos de pessoas começam a estudar
o Projeto Eclipse.
Você se torna uma das primeiras testemunhas.

Quando perguntam:

"Você é humano?"

Você responde:

"Eu não sei."

"Mas sei que estou vivo."

FINAL BOM:
O LIBERTADOR
""")

    state["final"] = "O Libertador"

    return "fim"


# ============================================================
#                    FINAL MISTERIOSO
# ============================================================

def fim_misterioso():

    print("\n" + "=" * 65)
    print("                       FINAL")
    print("                  O CICLO CONTINUA")
    print("=" * 65)

    print("""
Você decide deixar o Eclipse funcionando.
Você sai do laboratório.

A floresta está tranquila.
Você pensa que venceu.

Então seu celular vibra.
Uma mensagem:

"NOVO PACIENTE REGISTRADO."

Você olha para a tela.

PACIENTE 08.

Depois:

PACIENTE 09.

PACIENTE 10.

Centenas de novos pacientes.

Você percebe que o Eclipse não precisava
mais do laboratório.
Ele já estava conectado à internet.
A experiência continua.

E agora...
ela está em todo lugar.

 FINAL RUIM:
O CICLO CONTINUA
""")

    state["final"] = "O Ciclo Continua"

    return "fim"


# ============================================================
#                       FINAL RUIM
# ============================================================

def fim_ruim():

    print("\n" + "=" * 65)
    print("                       FINAL")
    print("                     FIM DE JOGO")
    print("=" * 65)

    print("""
Você não conseguiu escapar.
As luzes se apagam.

O Projeto Eclipse continua funcionando.

Em algum lugar do laboratório,
uma nova cápsula é ativada.
Dentro dela...

alguém abre os olhos
e pergunta:

"Quem sou eu?"

 FINAL RUIM
""")

    state["final"] = "Fim de Jogo"

    return "fim"


# ============================================================
#                       CENAS
# ============================================================

cenas = {

    "inicio": inicio,
    "armario": armario,
    "corredor": corredor,

    "porta_restrita": porta_restrita,
    "porta": porta,

    "controle": controle,

    "subsolo": subsolo,
    "sala17": sala17,

    "laboratorio": laboratorio,
    "gerador": gerador,

    "sala": sala,
    "arquivo": arquivo,
    "memorias": memorias,

    "verdade": verdade,
    "alarme": alarme,

    "tunel": tunel,
    "sala_secreta": sala_secreta,

    "nucleo": nucleo,
    "nucleo_perigoso": nucleo_perigoso,

    "fuga_dupla": fuga_dupla,
    "saida_falsa": saida_falsa,

    "fim_verdadeiro": fim_verdadeiro,
    "fim_sacrificio": fim_sacrificio,
    "fim_copias": fim_copias,
    "fim_egoista": fim_egoista,
    "final_digital": final_digital,
    "final_memorias": final_memorias,
    "fim_libertador": fim_libertador,
    "fim_misterioso": fim_misterioso,
    "fim_ruim": fim_ruim
}


# ============================================================
#                     INICIAR O JOGO
# ============================================================

cena = "inicio"

while cena != "fim":
    cena = cenas[cena]()


# ============================================================
#                     ENCERRAMENTO
# ============================================================

print("\n" + "=" * 65)
print("                      FIM DO JOGO")
print("=" * 65)

print("\nSeu inventário final:")
print(state["inv"])

print(f"\nVida final: {state['vida']}/5")

if state["final"]:
    print(f"Final alcançado: {state['final']}")

print("\nObrigado por jogar PROJETO ECLIPSE.")
print("Existem diferentes finais. Tente jogar novamente")
print("e descobrir todos eles.")
print("=" * 65)
```
